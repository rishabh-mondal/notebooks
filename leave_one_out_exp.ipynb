```python
import os
import shutil
from itertools import combinations
```

```python


# Define the airsheds and their paths
airsheds = {
    "delhi": {
        "images": "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/delhi_airshed/images",
        "labels": "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/delhi_airshed/labels"
    },
    "wb": {
        "images": "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/wb_small_airshed/images",
        "labels": "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/wb_small_airshed/labels"
    },
    "lucknow": {
        "images": "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/lucknow_airshed/images",
        "labels": "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/lucknow_airshed/labels"
    },
    "ahmedabad": {
        "images": "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/ahmedabad_10km_buffer/images",
        "labels": "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/ahmedabad_10km_buffer/labels"
    }
}

# Define base path for LOO datasets
loo_base_path = "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/leave_one_out"

# Function to copy files
def copy_files(src, dst):
    if not os.path.exists(dst):
        os.makedirs(dst)
    for file in os.listdir(src):
        shutil.copy(os.path.join(src, file), dst)

# Generate Leave-One-Out datasets
for test_zone in airsheds.keys():
    train_zones = [zone for zone in airsheds.keys() if zone != test_zone]
    split_path = os.path.join(loo_base_path, f"leave_out_{test_zone}")
    
    train_image_path = os.path.join(split_path, "train/images")
    train_label_path = os.path.join(split_path, "train/labels")
    test_image_path = os.path.join(split_path, "test/images")
    test_label_path = os.path.join(split_path, "test/labels")
    
    # Create directories
    os.makedirs(train_image_path, exist_ok=True)
    os.makedirs(train_label_path, exist_ok=True)
    os.makedirs(test_image_path, exist_ok=True)
    os.makedirs(test_label_path, exist_ok=True)
    
    # Copy training data
    for train_zone in train_zones:
        copy_files(airsheds[train_zone]["images"], train_image_path)
        copy_files(airsheds[train_zone]["labels"], train_label_path)
    
    # Copy testing data
    copy_files(airsheds[test_zone]["images"], test_image_path)
    copy_files(airsheds[test_zone]["labels"], test_label_path)
    
    print(f"Created Leave-One-Out dataset: leave_out_{test_zone}")
```
