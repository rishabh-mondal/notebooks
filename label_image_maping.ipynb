```python
import os
import random
from pathlib import Path

# Base paths
base_dir = Path("/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/sentinel/delhi_ncr")
image_dir = base_dir / "image"
label_dir = base_dir / "labels"
division_dir = base_dir / "devision_data"

# Set random seed
random.seed(42)

# Collect matching image-label pairs only
image_files = sorted([
    f for f in image_dir.glob("*.png")
    if (label_dir / f.with_suffix(".txt").name).exists()
])

print(f"✅ Found {len(image_files)} matching image-label pairs.")

# Split 80-20
random.shuffle(image_files)
split_index = int(0.8 * len(image_files))
train_files = image_files[:split_index]
test_files = image_files[split_index:]

# Function to create symlinks
def symlink_pairs(files, subset):
    img_out = division_dir / subset / "images"
    lbl_out = division_dir / subset / "labels"
    img_out.mkdir(parents=True, exist_ok=True)
    lbl_out.mkdir(parents=True, exist_ok=True)

    for img_path in files:
        stem = img_path.stem
        label_path = label_dir / f"{stem}.txt"

        img_symlink = img_out / img_path.name
        lbl_symlink = lbl_out / label_path.name

        try:
            if not img_symlink.exists():
                os.symlink(img_path.resolve(), img_symlink)
            if not lbl_symlink.exists():
                os.symlink(label_path.resolve(), lbl_symlink)
        except Exception as e:
            print(f"❌ Failed to symlink {img_path.name}: {e}")

# Symlink the pairs
symlink_pairs(train_files, "train")
symlink_pairs(test_files, "test")
```
